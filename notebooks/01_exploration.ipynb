# Etudes de CROP_YIELD_DF et de YIELD_DF

```python
display(crop_yield_df.head(5))
display(yield_df.head(5))
```

Pour fusionner les deux datasets nous avons besoins d'une colonne commune.
Dans `Crop_yield_df` nous avons **region** qui pourrai nous faire comprendre la situation géographique, l'homologue de ce dernier dans yield_df est "Area". Nous ne pouvons cependant faire une jointure en se basant sur ces deux variables.
Les seuls colonnes le permettant sont: **Crop** dans `Crop_yield_df`et **Item** dans `yield_df`. Nous allons donc renommer *Item* en *Crop* ce qui est plus parlant et logique pour la suite et faire une **Aggrégation afin d'obtenir les différentes mesures en moyenne**.

```python
#Renommer la colonne Item
yield_df.rename(columns={"Item": "Crop"}, inplace=True)

#Aggréger les données 
yield_df_agg = yield_df.groupby("Crop").agg({
    "hg/ha_yield": "mean", #Rendement moyen,
    "average_rain_fall_mm_per_year": "mean", #Pluviométrie moyenne
    "pesticides_tonnes": "mean", #Utilisation moyenne de pesticide,
    "avg_temp": "mean", #Température moyenne.
}).reset_index()

print("yield_df_agg")
display(yield_df.head())

crop_df = pd.merge(
    crop_yield_df,
    yield_df_agg,
    on="Crop",
    how="inner"
)

print("crop_df")
display(crop_df.head())
```

```python
print(f"Nombre de ligne de CROP_YIELD_DF {crop_yield_df.shape[0]}")
print(f"Nombre de ligne de CROP_DF {crop_df.shape[0]}")
```

```python
crop_df = pd.merge(
    crop_yield_df,
    yield_df_agg,
    on="Crop",
    how="left"
)

print("crop_df")
display(crop_df.head())

print(f"Nombre de ligne de CROP_YIELD_DF {crop_yield_df.shape[0]}")
print(f"Nombre de ligne de CROP_DF {crop_df.shape[0]}")
```

### Comparaison des jointures

La jointure entre `crop_yield_df` et `yield_df_agg` sur la colonne `Crop` produit :

- **333 497 lignes** avec une jointure `inner` ;
- **1 000 000 lignes** avec une jointure `left`, en conservant toutes les lignes de `crop_yield_df`.

La jointure `left` permet donc de ne perdre aucune observation du dataset principal `crop_yield_df`.

Nous avons donc soit 70% de lignes perdues en *inner join* soit 70% de valeurs nulles en *left join*.
Nous allons donc auditer nos catégories.

```python
crop_yield_items = set(crop_yield_df["Crop"].unique())
yield_items = set(yield_df["Crop"].unique())

print(f"Les éléments figurant dans crop_yield et ne figurant pas dans yield {crop_yield_items-yield_items}")
print(f"Les éléments figurant dans yield et ne figurant pas dans crop_yield {yield_items - crop_yield_items}")
```

Lors de l'analyse exploratoire croisée des jeux de données `crop_yield.csv` et `yield_df.csv`, nous avons identifié des divergences dans la dénomination de certaines cultures identiques. 

Ces différences de nomenclature s'expliquent par les sources des données (notamment la nomenclature spécifique de la FAO pour `yield_df.csv`). Afin de pouvoir fusionner correctement ces deux sources sans générer de valeurs manquantes (`NaN`) inutiles, nous devons standardiser les noms.

**Incohérences identifiées :**
*   **Le Soja :** Nommé `Soybean` (singulier) dans le jeu de données agronomique, mais `Soybeans` (pluriel) dans l'historique de rendement.
*   **Le Riz :** Nommé `Rice` dans le premier, et `Rice, paddy` (riz brut) dans le second.

**Stratégie de résolution :**
Nous appliquons un dictionnaire de correspondance (Mapping) pour aligner les valeurs de `yield_df.csv` sur le standard de `crop_yield.csv`. Cette étape est un prérequis indispensable avant de procéder à une jointure

```python
mapping_crop = {
    "Soybeans": "Soybean",
    "Rice, paddy": "Rice"
}

#faire le remplacement
yield_df_agg["Crop"] = yield_df_agg["Crop"].replace(mapping_crop)


crop_df = pd.merge(
    crop_yield_df,
    yield_df_agg,
    on="Crop",
    how="inner"
)

print("crop_df")
display(crop_df.head())

print(f"Nombre de ligne de CROP_YIELD_DF {crop_yield_df.shape[0]}")
print(f"Nombre de ligne de CROP_DF {crop_df.shape[0]}")
```

### Stratégie de Fusion : Préservation des données via Left Join

Après la phase de mapping, l'objectif est d'enrichir les données agronomiques détaillées de `crop_yield.csv` avec les moyennes historiques (climat et pesticides) issues de `yield_df.csv`.

**Le problème identifié :**
Les cultures d'orge (`Barley`) et de coton (`Cotton`), présentes dans `crop_yield.csv`, n'ont aucun équivalent dans les données historiques de la FAO (`yield_df.csv`). 

**Choix technique (Left Join vs Inner Join) :**
*   Une jointure stricte (`Inner Join`) nous aurait fait perdre **333 362 observations**, soit environ 33% de notre jeu de données, éliminant totalement l'orge et le coton de notre périmètre d'analyse.
*   **Décision :** Nous optons pour une jointure à gauche (`Left Join`) en utilisant `crop_yield.csv` comme table de référence. 

**Conséquences et prochaines étapes :**
Cette stratégie permet de conserver nos 1 000 000 d'observations intactes. Cependant, elle génère des valeurs manquantes (`NaN`) sur les variables historiques pour le coton et l'orge (environ 666 638 lignes ont trouvé une correspondance). Nous traiterons ces valeurs manquantes dans la prochaine phase via une stratégie d'imputation adaptée, afin de ne pas réduire le catalogue de cultures recommandables par notre API.

Pour éviter une colinéarité accrue sur les variables nous garderons que les colonnes `Crop` `Pesticides tonnes` du dataset aggrégé. Choix qui réduira nos NaN que nous remplirons par des 0 pour dire qu'on les a pas.

```python
crop_df = pd.merge(
    crop_yield_df,
    yield_df_agg[["Crop", "pesticides_tonnes"]],
    on="Crop",
    how="left"
)

crop_df.rename(columns={"pesticides_tonnes": "pesticides_tonnes_mean"}, inplace=True)

print("crop_df")
display(crop_df.head())

print(f"Nombre de ligne de CROP_YIELD_DF {crop_yield_df.shape[0]}")
print(f"Nombre de ligne de CROP_DF {crop_df.shape[0]}\n\n")

print("Nombre de valeurs nulles")
display(crop_df.isna().sum())

crop_df["pesticides_tonnes_mean"] = crop_df["pesticides_tonnes_mean"].fillna(0)
display(crop_df.head())



#Enregistrer le dataset nettoyé
crop_df.to_csv("../data/processed/final_dataset.csv", index=False)
```
