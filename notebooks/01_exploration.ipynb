### Définition des colonnes
Nous avons 10 variables regroupées 3 types:

- `Region` définie la région dans laquelle est faite la culture que décrit la ligne. 
Elle est de type `Nominale catégorielle`

- `Soil_Type` décrit le type de sol sur lequel se fait la culture décrite par la ligne. Elle est de type `Nominale catégorielle`

- `Crop` définie l'élémet cultivé sur le sol que décrit la ligne. 
Elle est de type `Nominale catégorielle`

- `Rainfall_mm` est la quantié de précipitation en milimètre sur la culture que décrit la ligne.
Elle est de type `Quantitative continue`

- `Temperature_Celsius` donne la température observée sur la culture qui est décrite par ligne.
Elle est de type `Quantitative continue`

- `Fertilizer_Used` donne l'information de l'usage ou du non usage de fertilisant sur la culture décrite sur la ligne.
Elle est de type `Nominale catégorielle binaire`

- `Irrigation_Used` donne l'information de l'usage ou du non usage d'un système d'irrigation sur la culture décrite sur la ligne.
Elle est de type `Nominale catégorielle binaire`

- `Weather_Condition` fournit l'information sur les conditions météoroliques sur la culture décrite par la ligne.
Elle est de type `Nominale catégorielle`

- `Days_to_Harvest` founit le nombre de jours qui se sont écoulés avant la récolte.
Elle est de type `Quantitative Discrète`

- `Yield_tons_per_hectare` donne le rendemenent en tonne et par heactare pour une culture donnée.
Elle est de type `Quantitative continue`

#### Trouver les différentes catégorie pour chaque variable

Nous avons précédemment vu que nous avions 6 variables , cette observation était uniquement visuelle.
Nous allons maintenant la vérifier par algorithmique.


```python
nbr_cat = 0
for col in crop_yield_df.columns:
    if crop_yield_df[col].dtype == "str" or crop_yield_df[col].dtype == "bool":
        nbr_cat+=1
        print(f"({col}) a {crop_yield_df[col].nunique()} occurences qui sont {crop_yield_df[col].unique()}\npour chaque occurence on a en nombre d'occurence : {crop_yield_df[col].value_counts()}\n\n")

print(f"Nous avons donc {nbr_cat} variables qualitatives catégorielles")
```
